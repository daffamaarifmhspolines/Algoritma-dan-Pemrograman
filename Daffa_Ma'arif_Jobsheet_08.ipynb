#Praktikum 04 - Fungsi dengan *args dan **kwargs

```python
def jumlahkan(*args):
  return sum(args)

def biodata(**kwargs):
  for key, value in kwargs.items():
    print(f"{key}: {value}")

print("Total:", jumlahkan(1, 2, 3, 4, 5))
biodata(Nama="Budi", Umur=20, Prodi="TRK")
```

#Praktikum 05 - Scope Variabel

```python
x = 10 # variabel global

def ubah():
  x = 5 # variabel lokal
  print("Nilai lokal:", x)

ubah()
print("Nilai global:", x)
```

#Praktikum 07 - Fungsi Decorator

```python
def decorator_sapa(fungsi):
  def wrapper():
    print("Sebelum fungsi dijalankan...")
    fungsi()
    print("Setelah fungsi dijalankan...")
  return wrapper

@decorator_sapa
def sapa():
  print("Halo, Mahasiswa Polines!")

sapa()
```

#Praktikum 08 - Studi Kasus: Menghitung Rata-rata Nilai Mahasiswa

```python
def hitung_rata(data):
  return sum(data) / len(data)

nilai = [80, 90, 85]
rata = hitung_rata(nilai)

print("Rata-rata nilai:", rata)
```

#Penugasan 04 - Fungsi dengan *args

```python
def jumlahkan_semua(*angka):
  total = sum(angka)
  print(f"Total dari {angka} adalah {total}")

jumlahkan_semua(1, 2, 3)
jumlahkan_semua(10, 20, 30, 40, 50)
jumlahkan_semua(5, 5, 5, 5, 5, 5, 5)
```

#Penugasan 05 - Fungsi dengan **kwargs

```python
def info_barang(**data):
  print("Info Barang:")
  for key, value in data.items():
    print(f"- {key}: {value}")

info_barang(Nama="Buku", Harga=15000, Stok=30)
```

#Penugasan 06 - Scope Variabel

```python
x = 100 # global

def ubah():
  x = 50 # lokal
  print(f"Nilai x di dalam fungsi (lokal): {x}")

ubah()
print(f"Nilai x di luar fungsi (global): {x}")
print("Penjelasan: Variabel 'x' di dalam fungsi adalah variabel lokal yang berbeda dari 'x' global.")
```

#Penugasan 08 - Fungsi Decorator

```python
def logger(fungsi):
  def wrapper():
    print("Memulai fungsi...")
    fungsi()
    print("Fungsi selesai dijalankan.")
  return wrapper

@logger
def halo():
  print("Halo, ini fungsi halo!")

halo()
```

#Penugasan 09 - Studi Kasus Nyata - Program Nilai

```python
def input_nilai():
  nilai_tugas = 80
  nilai_ujian = 70
  print(f"Nilai Tugas: {nilai_tugas}, Nilai Ujian: {nilai_ujian}")
  return nilai_tugas, nilai_ujian

def hitung_rata(list_nilai):
  return sum(list_nilai) / len(list_nilai)

def status_kelulusan(rata):
  if rata >= 60:
    return "Lulus"
  else:
    return "Tidak Lulus"

hitung_akhir = lambda tugas, ujian: (0.4 * tugas) + (0.6 * ujian)

tugas, ujian = input_nilai()
nilai_akhir = hitung_akhir(tugas, ujian)
print(f"Nilai Akhir: {nilai_akhir}")

semua_nilai = [tugas, ujian]
rata_sederhana = hitung_rata(semua_nilai)
print(f"Rata-rata sederhana: {rata_sederhana}")

status = status_kelulusan(nilai_akhir)
print(f"Status: {status}")
```
