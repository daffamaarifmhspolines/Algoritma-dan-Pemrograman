#Praktikum 01 - Bubble Sort

```python
def bubble_sort(data):
    n = len(data)
    for i in range(n):
        for j in range(0, n - i - 1):
            if data[j] > data[j + 1]:
                data[j], data[j + 1] = data[j + 1], data[j]
    return data

angka = [64, 34, 25, 12, 22, 11, 90]
print("Data sebelum diurutkan:", angka)
print("Data setelah Bubble Sort:", bubble_sort(angka))
```

#Praktikum 02 - Selection Sort

```python
def selection_sort(data):
    n = len(data)
    for i in range(n):
        min_index = i
        for j in range(i + 1, n):
            if data[j] < data[min_index]:
                min_index = j

        data[i], data[min_index] = data[min_index], data[i]
    return data

angka = [64, 25, 12, 22, 11]
print("Data sebelum diurutkan:", angka)
print("Data setelah Selection Sort:", selection_sort(angka))
```

#Praktikum 03 - Insertion Sort

```python
def insertion_sort(data):
    n = len(data)
    for i in range(1, n):
        key = data[i]
        j = i - 1

        while j >= 0 and data[j] > key:
            data[j + 1] = data[j]
            j -= 1
        data[j + 1] = key
    return data

angka = [12, 11, 13, 5, 6]
print("Data sebelum diurutkan:", angka)
print("Data setelah Insertion Sort:", insertion_sort(angka))
```

#Praktikum 04 - Studi Kasus: Mengurutkan Nilai Mahasiswa

```python
def bubble_sort_nilai(data):
    n = len(data)
    for i in range(n):
        for j in range(0, n - i - 1):
            if data[j][1] < data[j + 1][1]:
                data[j], data[j + 1] = data[j + 1], data[j]
    return data

mahasiswa = [("Andi", 80), ("Budi", 70), ("Cici", 90), ("Doni", 85)]
print("Data sebelum diurutkan:", mahasiswa)
print("Data setelah diurutkan (desc):", bubble_sort_nilai(mahasiswa))
```

#Praktikum 05 - Perbandingan Waktu Eksekusi Sorting

```python
import random
import time

data = [random.randint(1, 1000) for _ in range(500)]

for func in [bubble_sort, selection_sort, insertion_sort]:
    salinan = data.copy()
    start = time.time()
    func(salinan)
    end = time.time()
    print(f"{func.__name__} membutuhkan waktu {end - start:.5f} detik")
```

#Praktikum 06 - Sorting Data Barang

```python
def bubble_sort_key(data, key_index=0, reverse=False):
    n = len(data)
    for i in range(n):
        for j in range(0, n - i - 1):
            if reverse:
                if data[j][key_index] < data[j + 1][key_index]:
                    data[j], data[j + 1] = data[j + 1], data[j]
            else:
                if data[j][key_index] > data[j + 1][key_index]:
                    data[j], data[j + 1] = data[j + 1], data[j]
    return data

barang = [("Buku", 15000), ("Pensil", 5000), ("Tas", 75000), ("Pulpen", 10000)]

print("Urut berdasarkan nama (asc):", bubble_sort_key(barang.copy(), 0))
print("Urut berdasarkan harga (desc):", bubble_sort_key(barang.copy(), 1, reverse=True))
```

#Penugasan 03 - Eksperimen Bubble Sort (Descending)

```python
def bubble_sort_descending(arr):
    n = len(arr)
    for i in range(n):
        for j in range(0, n - i - 1):
            if arr[j] < arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
    return arr

angka = [64, 34, 25, 12, 22, 11, 90]
print("Descending Bubble Sort:", bubble_sort_descending(angka))
```

#Penugasan 04 - Eksperimen Selection Sort (Max at End)

```python
def selection_sort_max(arr):
    n = len(arr)
    for i in range(n):
        max_index = 0
        last_unsorted_index = n - 1 - i

        for j in range(1, last_unsorted_index + 1):
            if arr[j] > arr[max_index]:
                max_index = j

        arr[last_unsorted_index], arr[max_index] = arr[max_index], arr[last_unsorted_index]
    return arr

angka = [64, 25, 12, 22, 11]
print("Selection Sort (Max to End):", selection_sort_max(angka))
```

#Penugasan 05 - Eksperimen Insertion Sort (Data String)

```python
def insertion_sort_string(arr):
    n = len(arr)
    for i in range(1, n):
        key = arr[i]
        j = i - 1
        while j >= 0 and arr[j] > key:
            arr[j + 1] = arr[j]
            j -= 1
        arr[j + 1] = key
    return arr

mahasiswa = ["Python", "Anaconda", "Cobra", "Viper", "Taipan"]
print("Insertion Sort (String):", insertion_sort_string(mahasiswa))
```

#Penugasan 07 - Analisis Kompleksitas

```python
import random

def bubble_sort_count(arr):
    count = 0
    n = len(arr)
    for i in range(n):
        for j in range(0, n - i - 1):
            count += 1
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
    return count

def selection_sort_count(arr):
    count = 0
    n = len(arr)
    for i in range(n):
        min_index = i
        for j in range(i + 1, n):
            count += 1
            if arr[j] < arr[min_index]:
                min_index = j
        arr[i], arr[min_index] = arr[min_index], arr[i]
    return count

def insertion_sort_count(arr):
    count = 0
    n = len(arr)
    for i in range(1, n):
        key = arr[i]
        j = i - 1
        while j >= 0 and arr[j] > key:
            count += 1
            arr[j + 1] = arr[j]
            j -= 1
        count += 1
        arr[j + 1] = key
    return count

sizes = [10, 50, 100]
print("Analisis Jumlah Perbandingan:")
print(f"{'Ukuran':<10} {'Bubble':<10} {'Selection':<10} {'Insertion':<10}")

for size in sizes:
    data = [random.randint(1, 1000) for _ in range(size)]
    c_bubble = bubble_sort_count(data.copy())
    c_selection = selection_sort_count(data.copy())
    c_insertion = insertion_sort_count(data.copy())
    print(f"{size:<10} {c_bubble:<10} {c_selection:<10} {c_insertion:<10}")
```
