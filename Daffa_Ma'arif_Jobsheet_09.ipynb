#Penugasan 01 - Eksperimen dengan Modul math

```python
import math

def luas_lingkaran(jari_jari):
  return math.pi * (jari_jari ** 2)

print(f"Luas lingkaran jari-jari 7: {luas_lingkaran(7):.2f}")
print(f"Luas lingkaran jari-jari 10: {luas_lingkaran(10):.2f}")
```

#Penugasan 04 - Modul Buatan Sendiri

```python
%%writefile konversi.py
def celcius_ke_fahrenheit(c):
  return (c * 9/5) + 32

def fahrenheit_ke_celcius(f):
  return (f - 32) * 5/9
```

```python
import konversi

c = 100
f = 212
print(f"{c}C adalah {konversi.celcius_ke_fahrenheit(c)}F")
print(f"{f}F adalah {konversi.fahrenheit_ke_celcius(f)}C")
```

#Penugasan 06 - Studi Kasus Mini

```python
%%writefile laporan.py
def tampil_skor(skor, total_soal):
  print("\n--- Kuis Selesai ---")
  print(f"Skor Anda: {skor} benar dari {total_soal} soal.")
  persentase = (skor / total_soal) * 100
  print(f"Nilai: {persentase:.2f}")
```

```python
import random
import laporan

skor = 0
total_soal = 3

print("=== Kuis Penjumlahan Acak ===")
for i in range(total_soal):
  a = random.randint(1, 20)
  b = random.randint(1, 20)
  jawaban_benar = a + b

  jawaban_user = int(input(f"Soal {i+1}: {a} + {b} = "))

  if jawaban_user == jawaban_benar:
    print("Benar!")
    skor += 1
  else:
    print(f"Salah. Jawaban yang benar adalah {jawaban_benar}")

laporan.tampil_skor(skor, total_soal)
```
